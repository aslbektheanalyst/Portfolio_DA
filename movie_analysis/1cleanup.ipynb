```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("imdb_top_1000.csv")
```

```python
df.shape
df.info()
```

```python
#checking if a movie is referenced more than once
df['Poster_Link'].is_unique
```

```python
#fixing gross column, dropping empty ones

df["Gross"] = pd.to_numeric(
    df["Gross"].str.replace(",", "", regex=False),
    errors="coerce"
).astype("Int64")

df.dropna(subset="Gross", inplace=True)
```

```python
#fixing runtime column

df["Runtime"] = (
    df["Runtime"]
    .str.replace(" min", "", regex=False)
    .astype("Int64")
)
```

```python
#fixing release year column

df["Released_Year"] = pd.to_numeric(df["Released_Year"], errors="coerce").astype("Int64")
df.dropna(subset='Released_Year', inplace = True)
```

```python
df.describe()
```

```python
#adding another column based on the gross column

bins = [0, 10_000_000, 50_000_000, 100_000_000, float("inf")]
labels = ["Low", "Moderate", "High", "Blockbuster"]

df["gross_category"] = pd.cut(
    df["Gross"],
    bins=bins,
    labels=labels
)
```

```python
print("Rows:", len(df))
print("\nMissing values:")
print(df.isna().sum())
```

```python
df.to_csv("cleaned_data.csv", index=False)
```
