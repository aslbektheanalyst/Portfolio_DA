```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('ready_housesales_data.csv')
```

```python
"""
ACCORDING TO THE WEBSITE THAT PROVIDED THIS DATASET, THE FILE IS ALREADY CLEANED UP AND READY TO BE 
USED FOR MACHINE LEARNING AND VISUALISATION. I JUST NEED TO ADD A FEW MEASURE COLUMNS AND IT WILL BE READY TO USE FOR MY PROJECT TOO
"""
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
current_year = 2026
df['house_age'] = current_year - df['yr_built']
```

```python
df['has_basement'] = df['sqft_basement'].apply(lambda x: True if x > 0 else False)
```

```python
df['is_renovated'] = df['yr_renovated'].apply(lambda x: True if x > 0 else False)
```

```python
def calculate_estimated_price(row):

    base_price = row['sqft_living'] * row['price_per_sqft']

    condition_multiplier = 1 + (row['condition'] - 3) * 0.05
    
    age_discount = max(0, row['house_age'] * 0.002)

    waterfront_bonus = 0.30 if row['waterfront'] == 1 else 0.0

    adjusted_price = base_price * condition_multiplier * (1 - age_discount) * (1 + waterfront_bonus)

    return round(adjusted_price, 2)


df['estimate_price'] = df.apply(calculate_estimated_price, axis=1)
```
