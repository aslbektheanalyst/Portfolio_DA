```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('housesales_wa.csv')
```

```python
"""
ACCORDING TO THE WEBSITE THAT PROVIDED THIS DATASET, THE FILE IS ALREADY CLEANED UP AND READY TO BE 
USED FOR MACHINE LEARNING AND VISUALISATION. I JUST NEED TO ADD A FEW MEASURE COLUMNS AND IT WILL BE READY TO USE FOR MY PROJECT TOO
"""
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
current_year = 2026
df['house_age'] = current_year - df['yr_built']
```

```python
df['has_basement'] = df['sqft_basement'].apply(lambda x: True if x > 0 else False)
```

```python
df['is_renovated'] = df['yr_renovated'].apply(lambda x: True if x > 0 else False)
```

```python
df.describe()
```

```python
print((df["price"] < 10_000).sum())
print((df["price"] == 0).sum())
```

```python
metrics = ["view", "floors", "waterfront", "yr_renovated"]

normal_price = (
    df[df["price"] >= 50_000]
    .groupby(metrics)["price"]
    .transform("mean")
)

error_mask = df["price"] < 50_000

df.loc[error_mask, "price"] = normal_price[error_mask]
```
