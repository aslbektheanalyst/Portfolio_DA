```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('sales_transactions.csv')
```

```python
df.iloc[5].to_frame("sample")
```

```python
#removing unnecessary columns
df2 = df[[
    'Order_ID',
    'Customer_ID',
    'Customer_Name',
    'Customer_Age',
    'Order_Date',
    'Order_Time',
    'Store_Name',
    'Country',
    'City',
    'Product_ID',
    'Product_Name',
    'Product_Category',
    'Quantity',
    'Unit_Price',
    'Discount_Percentage',
    'Sales_Amount',
    'Cost_Amount',
    'Payment_Method',
    'Sales_Representative',
    'Promotion_Code',
    'Order_Year'
]].copy()
```

```python
df2.describe(include='all')
```

```python
df2.shape
```

```python
df2.columns = df2.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
print(df2.duplicated().sum())
df2 = df2.drop_duplicates()
```

```python
df2 = df2.replace(["N/A", "NA", "null", "-", ""], np.nan)
df2.isna().sum()
```

```python
#filling the empty cells in columns

df2["customer_name"] = df2["customer_name"].fillna("Unknown")
df2["customer_age"] = df2["customer_age"].fillna(df2["customer_age"].median())
df2["product_name"] = df2["product_name"].fillna("Unknown Product")
df2["payment_method"] = df2["payment_method"].fillna("Unknown")
```

```python
df2.dtypes
```

```python
#correcting dtypes

df2["customer_age"] = df2["customer_age"].astype(int)
df2["order_date"] = pd.to_datetime(df2["order_date"])
df2["order_time"] = pd.to_datetime(df2["order_time"]).dt.time
```

```python
#checking impossible values in numerical columns
df2.select_dtypes("number").describe()
```

```python
#checking to see why quantity has below zero values, if the main df has returning data or not

df2[df2["quantity"] <= 0].head(5)
```

```python
df2 = df2[df2["quantity"] > 0].copy()
```

```python
print(df2["payment_method"].unique().tolist())
print(df2["product_category"].unique().tolist())
print(df2["sales_representative"].unique().tolist())
```

```python
#making all lowercase for those 2 columns

df2["payment_method"] = df2["payment_method"].str.lower()
df2["product_category"] = df2["product_category"].str.lower()
```

```python
#same order ids are not by mistake, nor duplicate orders

df2[df2["order_id"].duplicated(keep=False)].sort_values("order_id").head(4)
```
